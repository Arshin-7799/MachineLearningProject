```python
from pathlib import Path
import pandas as pd
import numpy as np
DATA_DIR=Path("../../ecommerce_dataset")
print("pandas version:",pd.__version__)
print("data folder:",DATA_DIR.resolve())
```

```python
import numpy as np
users=pd.read_csv(
    DATA_DIR/"users.csv"
)
users.info()
users.describe()
```

```python
import numpy as np
products=pd.read_csv(
    DATA_DIR/"products.csv"
)
products.info()
products.describe()
```

```python
import numpy as np
orders=pd.read_csv(
    DATA_DIR/"orders.csv"
)
orders.info()
orders.describe()
```

```python
import numpy as np
order_items=pd.read_csv(
    DATA_DIR/"order_items.csv"
)
order_items.info()
order_items.describe()
```

```python
import numpy as np
events=pd.read_csv(
    DATA_DIR/"events.csv"
)
events.info()
events.describe()
```

```python
import numpy as np
reviews=pd.read_csv(
    DATA_DIR/"reviews.csv"
)
reviews.info()
reviews.describe()
```

```python
print(users["signup_date"].head())
print(orders["order_date"].head())
print(reviews["review_date"].head())
print(events["event_timestamp"].head())
```

```python
date_format = "%Y-%m-%dT%H:%M:%S.%f"
timestamp_checks = pd.Series({
    "users": pd.to_datetime(
        users["signup_date"],
        errors="coerce"
    ).isna().sum(),
    "orders": pd.to_datetime(
        orders["order_date"],
        format=date_format,
        errors="coerce"
    ).isna().sum(),
    "reviews": pd.to_datetime(
        reviews["review_date"],
        format=date_format,
        errors="coerce"
    ).isna().sum(),
    "events": pd.to_datetime(
        events["event_timestamp"],
        format=date_format,
        errors="coerce"
    ).isna().sum()
})
timestamp_checks.to_frame("invalid or missing")
```

```python
date_format = "%Y-%m-%dT%H:%M:%S.%f"
signup_format="%Y-%m-%d"
users["signup_date"] = pd.to_datetime(
    users["signup_date"],
    format=signup_format,    
)
users.dtypes
```

```python
date_format = "%Y-%m-%dT%H:%M:%S.%f"
orders["order_date"] = pd.to_datetime(
    orders["order_date"],
    format=date_format,
)
orders.dtypes
```

```python
date_format = "%Y-%m-%dT%H:%M:%S.%f"
reviews["review_date"] = pd.to_datetime(
    reviews["review_date"],
    format=date_format,
)
reviews.dtypes
```

```python
date_format = "%Y-%m-%dT%H:%M:%S.%f"
events["event_timestamp"] = pd.to_datetime(
    events["event_timestamp"],
    format=date_format,
)
events.dtypes
```

```python
users["date"] = users["signup_date"].dt.floor("D")
display(users[["signup_date","date"]].head())
orders["date"] = orders["order_date"].dt.floor("D")
display(orders[["order_date","date"]].head())
reviews["date"] = reviews["review_date"].dt.floor("D")
display(reviews[["review_date","date"]].head())
events["date"] = events["event_timestamp"].dt.floor("D")
display(events[["event_timestamp","date"]].head())
```

```python
product_summary = order_items.groupby("product_id").agg(
    total_quantity=("quantity", "sum"),
    average_price=("item_price", "mean"),
    total_sales=("item_total", "sum")
).reset_index()

display(product_summary)
```

```python
orders_daily = orders.groupby(
    ["date"], as_index=False
).agg(
    total_orders=("order_id", "nunique"),
    total_sales=("total_amount", "sum"),
    average_order_value=("total_amount", "mean")
)

display(orders_daily)
```

```python
events_daily = events.groupby(
    ["date"], as_index=False
).agg(
    total_events=("event_id", "nunique"),
    unique_users=("user_id", "nunique"),
    unique_products=("product_id", "nunique")
)

display(events_daily)
```

```python
reviews_daily = reviews.groupby(
    ["date"], as_index=False
).agg(
    total_reviews=("review_id", "nunique"),
    average_rating=("rating", "mean")
)

display(reviews_daily)
```

```python
daily_summary = orders_daily.merge(
    events_daily,
    on="date",
    how="outer",
    validate="one_to_one"
)

display(daily_summary)
```

```python
daily_summary = daily_summary.merge(
    reviews_daily,
    on="date",
    how="outer",
    validate="one_to_one"
)

display(daily_summary)
```

```python
from importlib.util import find_spec
from pathlib import Path
import pandas as pd
import plotly
import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
from plotly.subplots import make_subplots
DATA_DIR = Path("archive")
OUTPUT_DIR = Path("output")
OUTPUT_DIR.mkdir(exist_ok=True)
CHART_CONFIG = {
    "displaylogo": False,
    "responsive": True,
}
print("Plotly version:", plotly.__version__)
```

```python
order_status_summary = (
    orders.groupby("order_status", as_index=False)
    .agg(
        total_orders=("order_id", "nunique"),
        total_sales=("total_amount", "sum")
    )
    .sort_values("total_orders", ascending=False)
)

display(order_status_summary)
```

```python
import plotly.express as px

fig = px.pie(
    order_status_summary,
    names="order_status",
    values="total_orders",
    title="Order Status Distribution",
    hole=0.45
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    width=850,
    height=550
)

fig.update_traces(
    textinfo="percent+label",
    hovertemplate="<b>%{label}</b><br>Orders: %{value:,}<br>Percentage: %{percent}<extra></extra>"
)

fig.show()
```

```python
# Merge order items with product information

category_sales = order_items.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

# Calculate sales by category

category_sales = (
    category_sales
    .groupby("category", as_index=False)
    .agg(
        total_sales=("item_total", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

display(category_sales)
```

```python
ecommerce_dataset = order_items.merge(
    products[["product_id", "product_name", "category"]],
    on="product_id",
    how="left"
)

display(ecommerce_dataset.head())
ecommerce_dataset["total_quantity"] = ecommerce_dataset["quantity"]
ecommerce_dataset["total_items_sales"] = ecommerce_dataset["item_total"]
```

```python
import plotly.express as px
category_sales = ecommerce_dataset.groupby("category", as_index=False).agg(
    total_quantity=("total_quantity", "sum")
)
bar_chart= px.bar(
    category_sales,
    x="category",
    y="total_quantity",
    color="category",
    text_auto=True,
    color_discrete_sequence=px.colors.qualitative.Set2,
    title="Total Quantity Sold by Category",

    labels={
        "category": "Category",
        "total_quantity": "Total Quantity Sold"

    },

    template="plotly_white"
)
bar_chart.show(config=CHART_CONFIG)
```

```python
import plotly.express as px

monthly_sales = (
    orders.assign(month=orders["order_date"].dt.to_period("M").astype(str))
    .groupby("month", as_index=False)
    .agg(total_sales=("total_amount", "sum"))
    .sort_values("month")
)

fig = px.bar(
    monthly_sales,
    x="month",
    y="total_sales",
    title="Monthly Sales Performance",
    labels={
        "month": "Month",
        "total_sales": "Total Sales"
    },
    color="month",
    color_discrete_sequence=px.colors.qualitative.Set2,
    text_auto=".2s"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    width=1100,
    height=550,
    xaxis=dict(
        tickangle=-45,
        showgrid=False
    ),
    yaxis=dict(
        showgrid=True,
        tickformat=",.0f"
    ),
    margin=dict(
        l=60,
        r=40,
        t=80,
        b=100
    )
)

fig.update_traces(
    textposition="outside",
    hovertemplate="<b>%{x}</b><br>Total Sales: ₹%{y:,.2f}<extra></extra>"
)

fig.show()
```

```python
category_sales = (
    order_items
    .merge(products[["product_id", "category"]], on="product_id", how="left")
    .groupby("category", as_index=False)
    .agg(
        total_sales=("item_total", "sum"),
        total_quantity=("quantity", "sum")
    )
    .sort_values("total_sales", ascending=False)
)

display(category_sales)
```

```python
top_products = (
    order_items
    .merge(
        products[["product_id", "product_name"]],
        on="product_id",
        how="left"
    )
    .groupby("product_name", as_index=False)
    .agg(
        total_sales=("item_total", "sum"),
        total_quantity=("quantity", "sum")
    )
    .sort_values("total_sales", ascending=False)
    .head(10)
)

display(top_products)
```

```python
import plotly.express as px

fig = px.bar(
    top_products.sort_values("total_sales"),
    x="total_sales",
    y="product_name",
    orientation="h",
    title="Top 10 Products by Sales",
    labels={
        "product_name": "Product",
        "total_sales": "Total Sales"
    },
    color="product_name",
    color_discrete_sequence=px.colors.qualitative.Set2,
    text_auto=".2s"
)

fig.update_layout(
    template="plotly_white",
    title_x=0.5,
    width=1100,
    height=600,
    xaxis=dict(
        title="Total Sales",
        showgrid=True,
        tickformat=",.0f"
    ),
    yaxis=dict(
        title="Product",
        showgrid=False
    )
)

fig.update_traces(
    textposition="outside",
    hovertemplate="<b>%{y}</b><br>Total Sales: ₹%{x:,.2f}<extra></extra>"
)

fig.show()
```

```python
price_line = order_items.merge(
    products[["product_id", "category"]],
    on="product_id",
    how="left"
)

price_line = price_line.groupby(
    "category", as_index=False
).agg(
    average_price=("item_price", "mean")
)

line = px.line(
    price_line,
    x="category",
    y="average_price",
    markers=True,
    title="Average Item Price by Category",
    labels={
        "category": "Category",
        "average_price": "Average Item Price"
    },
    hover_data=["category", "average_price"],
    template="plotly_dark"
)

line.update_traces(
    line=dict(width=3),
    marker=dict(size=8)
)

line.show()
```

```python
orders_line = reviews.groupby(
    "order_id", as_index=False
).agg(
    number_of_reviews=("review_id", "nunique")
)

orders_line = orders_line.head(20)

line_chart = px.line(
    orders_line,
    x="order_id",
    y="number_of_reviews",
    markers=True,
    title="Number of Reviews by Order",
    labels={
        "order_id": "Order ID",
        "number_of_reviews": "Number of Reviews"
    },
    hover_data=["order_id", "number_of_reviews"],
    template="plotly_dark"
)

line_chart.update_traces(
    line=dict(width=3),
    marker=dict(size=8)
)

line_chart.show()
```

```python
import plotly.graph_objects as go
from plotly.subplots import make_subplots

# Create dashboard
dashboard = make_subplots(
    rows=3,
    cols=3,

    specs=[
        [{"type": "xy"}, {"type": "domain"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "xy"}, {"type": "xy"}],
        [{"type": "xy"}, {"type": "domain"}, {"type": "xy"}]
    ],

    subplot_titles=[
        "Monthly Sales Trend",
        "Order Status Distribution",
        "Sales By Category",
        "Product Price Distribution",
        "Total Quantity vs Average Item Sales",
        "Rating Distribution by Category",
        "",
        "Event Type Distribution",
        "Average Item Price by Category"


    ],
    vertical_spacing=0.12,
    horizontal_spacing=0.08
)

dashboard.add_trace(
    go.Scatter(
        x=monthly_sales["month"],
        y=monthly_sales["total_sales"],
        mode="lines+markers",
        name="Monthly Sales"
    ),
    row=1, col=1
)
dashboard.add_trace(
    go.Pie(
        labels=order_status_summary["order_status"],
        values=order_status_summary["total_orders"],
        hole=0.45
    ),
    row=1, col=2
)
dashboard.add_trace(
    go.Bar(
        x=category_sales["category"],
        y=category_sales["total_sales"],
        name="Category Sales"
    ),
    row=1, col=3
)
dashboard.add_trace(
    go.Histogram(
        x=products["price"],
        name="Product Price"
    ),
    row=2, col=1
)
dashboard.add_trace(
    go.Scatter(
        x=scatter_data["total_quantity"],
        y=scatter_data["average_item_sales"],
        mode="markers",
        name="Quantity vs Sales"
    ),
    row=2, col=2
)
for category in products["category"].unique():
    data = products[products["category"] == category]
    dashboard.add_trace(
        go.Box(
            y=data["rating"],
            name=category,
            boxpoints="outliers"
        ),
        row=2, col=3
    )
dashboard.add_trace(
    go.Pie(
        labels=event_type_summary["event_type"],
        values=event_type_summary["count"],
        hole=0.45
    ),
    row=3, col=2
)
dashboard.add_trace(
    go.Scatter(
        x=price_line["category"],
        y=price_line["average_price"],
        mode="lines+markers",
        name="Average Price"
    ),
    row=3, col=3
)
dashboard.update_layout(
    title="E-Commerce Data Analytics Dashboard",
    title_x=0.5,
    template="plotly_white",
    width=1500,
    height=1100,
    showlegend=False
)

dashboard.show()
```
